# Gold Layer Creation : DataWarehouseAnalytics

This notebook builds the **Gold Layer** of the DataWarehouseAnalytics project.

The Gold Layer contains **business-ready dimension and fact tables** optimized for reporting, BI dashboards and analytical workloads.

Tables created:
- dim_customers
- dim_products
- fact_sales


```sql
%sql
/*
Create Gold Layer Tables
---------------------------
Context:
    Dimension tables store descriptive entities such as customers and products.
    The fact table stores measurable sales transactions.

Databricks Notes:
    Tables are created as Delta tables to support reliable storage,
    schema management and efficient analytical queries.

*/

CREATE SCHEMA IF NOT EXISTS datawarehouseanalytics_gold;

USE datawarehouseanalytics_gold;

-- Customer dimension: contains customer attributes used for segmentation and analysis.
CREATE TABLE IF NOT EXISTS dim_customers (
    customer_key INT,
    customer_id INT,
    customer_number STRING,
    first_name STRING,
    last_name STRING,
    country STRING,
    marital_status STRING,
    gender STRING,
    birthdate DATE,
    create_date DATE
)
USING DELTA;

-- Product dimension: contains product details used to analyze sales by category, subcategory and product line.
CREATE TABLE IF NOT EXISTS dim_products (
    product_key INT,
    product_id INT,
    product_number STRING,
    product_name STRING,
    category_id STRING,
    category STRING,
    subcategory STRING,
    maintenance STRING,
    cost INT,
    product_line STRING,
    start_date DATE
)
USING DELTA;

-- Sales fact table: contains transactional sales metrics linked to customer and product dimensions.
CREATE TABLE IF NOT EXISTS fact_sales (
    order_number STRING,
    product_key INT,
    customer_key INT,
    order_date DATE,
    shipping_date DATE,
    due_date DATE,
    sales_amount INT,
    quantity TINYINT,
    price INT
)
USING DELTA;
```

```sql
%sql
/*

Validate Loaded Tables

Purpose:
    Verify that all gold layer tables were loaded successfully.

*/

SELECT 'dim_customers' AS table_name, COUNT(*) AS row_count
FROM datawarehouseanalytics_gold.dim_customers

UNION ALL

SELECT 'dim_products' AS table_name, COUNT(*) AS row_count
FROM datawarehouseanalytics_gold.dim_products

UNION ALL

SELECT 'fact_sales' AS table_name, COUNT(*) AS row_count
FROM datawarehouseanalytics_gold.fact_sales;
```
