# 02 -Database Exploration
Explore the structure of the gold layer by checking available tables,column names, data types and schema details.

Focus:
    This step helps understand how the analytical dataset is organized
    before performing business-level EDA.

```sql
%sql
/*

Database Exploration
--------------------
Purpose:
    Explore the structure of the gold layer by checking available tables,
    column names, data types and schema details.

Focus:
    This step helps understand how the analytical dataset is organized
    before performing business-level EDA.

Tables Used:
    - information_schema.tables
    - information_schema.columns

*/

-- Set the working schema for this analysis.
USE datawarehouseanalytics_gold;
```

```sql
%sql
-- List all tables available in the gold layer.
SELECT 
    table_catalog,
    table_schema,
    table_name,
    table_type
FROM information_schema.tables
WHERE table_schema = 'datawarehouseanalytics_gold';
```

```sql
%sql
-- Inspect column metadata for all gold layer tables.
SELECT 
    table_name,
    column_name,
    data_type,
    is_nullable,
    ordinal_position
FROM information_schema.columns
WHERE table_schema = 'datawarehouseanalytics_gold'
  AND table_name IN ('dim_customers', 'dim_products', 'fact_sales')
ORDER BY table_name, ordinal_position;
```
