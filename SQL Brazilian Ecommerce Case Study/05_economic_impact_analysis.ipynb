#Economic Impact Analysis

This notebook analyzes money movement in the e-commerce dataset using payment value, order price, and freight value.

Focus areas:
- Payment value growth from 2017 to 2018
- Total and average order price by state
- Total and average freight value by state

```sql
/*
Payment Value Growth: 2017 vs 2018

Purpose:
    Calculate the percentage increase in order payment value from 2017 to 2018.
    Only January to August are compared to keep both years on the same time range.
*/
WITH yearly_payment AS (
    SELECT
        YEAR(o.order_purchase_timestamp) AS order_year,
        SUM(p.payment_value) AS total_payment_value
    FROM ecommerce_analysis.payments p
    LEFT JOIN ecommerce_analysis.orders o
        ON p.order_id = o.order_id
    WHERE MONTH(o.order_purchase_timestamp) BETWEEN 1 AND 8
      AND YEAR(o.order_purchase_timestamp) IN (2017, 2018)
    GROUP BY YEAR(o.order_purchase_timestamp)
)

SELECT
    MAX(CASE WHEN order_year = 2017 THEN total_payment_value END) AS payment_2017,
    MAX(CASE WHEN order_year = 2018 THEN total_payment_value END) AS payment_2018,
    ROUND(
        (
            MAX(CASE WHEN order_year = 2018 THEN total_payment_value END)
            - MAX(CASE WHEN order_year = 2017 THEN total_payment_value END)
        ) * 100.0
        / MAX(CASE WHEN order_year = 2017 THEN total_payment_value END),
        2
    ) AS percentage_increase
FROM yearly_payment;



```

```sql
/*
Order Price by State

Purpose:
    Calculate total and average item price by customer state.
    This shows how order value differs across regions.
*/
SELECT
    c.customer_state AS state,
    ROUND(SUM(oi.price), 2) AS total_order_price,
    ROUND(AVG(oi.price), 2) AS avg_order_price
FROM ecommerce_analysis.order_items oi
LEFT JOIN ecommerce_analysis.orders o
    ON oi.order_id = o.order_id
LEFT JOIN ecommerce_analysis.customer c
    ON o.customer_id = c.customer_id
GROUP BY c.customer_state
ORDER BY total_order_price DESC;
```

```sql
/*
Freight Value by State

Purpose:
    Calculate total and average freight value by customer state.
    This helps compare logistics cost across regions.
*/
SELECT
    c.customer_state AS state,
    ROUND(SUM(oi.freight_value), 2) AS total_freight_value,
    ROUND(AVG(oi.freight_value), 2) AS avg_freight_value
FROM ecommerce_analysis.order_items oi
LEFT JOIN ecommerce_analysis.orders o
    ON oi.order_id = o.order_id
LEFT JOIN ecommerce_analysis.customer c
    ON o.customer_id = c.customer_id
GROUP BY c.customer_state
ORDER BY total_freight_value DESC;
```
