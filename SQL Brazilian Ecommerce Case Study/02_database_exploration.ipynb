#Database Exploration

This notebook explores the basic structure and characteristics of the dataset.

Focus areas:
- Customer table column metadata
- Order date range
- Initial understanding of available tables and fields

```sql
/*
Customer Table Structure

Purpose:
    Inspect the column names, data types, and nullability of the customers table.
    This helps understand the structure of customer-related fields before analysis.
*/
SELECT
    table_name,
    column_name,
    data_type,
    is_nullable
FROM information_schema.columns
WHERE table_schema = 'ecommerce_analysis'
  AND table_name = 'customer';
```

```sql
%sql
/*
Order Purchase Date Range

Purpose:
    Identify the first and latest order purchase dates and calculate
    the total time range covered by the orders dataset.
*/

SELECT
    MIN(order_purchase_timestamp) AS first_purchase_date,
    MAX(order_purchase_timestamp) AS latest_purchase_date,
    ROUND(
        MONTHS_BETWEEN(
            MAX(order_purchase_timestamp),
            MIN(order_purchase_timestamp)
        ),
        2
    ) AS time_range_months,
    ROUND(
        MONTHS_BETWEEN(
            MAX(order_purchase_timestamp),
            MIN(order_purchase_timestamp)
        ) / 12,
        2
    ) AS time_range_years
FROM ecommerce_analysis.orders;
```
