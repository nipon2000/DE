#Create Schema and Validate Tables

This notebook sets up the schema for the e-commerce SQL analysis project and validates the uploaded source tables.

The dataset contains customer, order, product, payment, review, seller, and geolocation data.  
The row count validation confirms that all source tables were successfully created in Databricks before starting EDA.

```sql
%sql
/*
Create Project Schema

Purpose:
    Create a dedicated schema for the e-commerce analysis project.
    All uploaded CSV tables are stored under this schema for organized querying.
*/

CREATE SCHEMA IF NOT EXISTS ecommerce_analysis;
```

```sql
%sql
/*
Validate Uploaded Source Tables

Purpose:
    Check the row count of each uploaded source table.
    This confirms that all CSV files were successfully loaded into Databricks.
*/

SELECT 'customers' AS table_name, COUNT(*) AS row_count FROM ecommerce_analysis.customer
UNION ALL
SELECT 'geolocation', COUNT(*) FROM ecommerce_analysis.geolocation
UNION ALL
SELECT 'order_items', COUNT(*) FROM ecommerce_analysis.order_items
UNION ALL
SELECT 'order_reviews', COUNT(*) FROM ecommerce_analysis.order_reviews
UNION ALL
SELECT 'orders', COUNT(*) FROM ecommerce_analysis.orders
UNION ALL
SELECT 'payments', COUNT(*) FROM ecommerce_analysis.payments
UNION ALL
SELECT 'products', COUNT(*) FROM ecommerce_analysis.products
UNION ALL
SELECT 'sellers', COUNT(*) FROM ecommerce_analysis.sellers
ORDER BY table_name;
```

```sql
%sql
/*
Inspect Table Structures

Purpose:
    Review column names, data types, and nullability for all source tables.
    This helps understand the database structure before writing analysis queries.
*/

SELECT
    table_name,
    column_name,
    data_type,
    is_nullable,
    ordinal_position
FROM information_schema.columns
WHERE table_schema = 'ecommerce_analysis'
  AND table_name IN (
      'customer',
      'geolocation',
      'order_items',
      'order_reviews',
      'orders',
      'payments',
      'products',
      'sellers'
  )
ORDER BY table_name, ordinal_position;
```
